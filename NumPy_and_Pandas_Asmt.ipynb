```python
# 2. NumPy Array Operations

# Task 1: Create a 1D NumPy Array

import numpy as np
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print("Week 1 Temperatures:")
print(temperatures_w1)
```

```python
# Task 2: Inspect Array Properties

print("Shape:", temperatures_w1.shape)
print("Data Type:", temperatures_w1.dtype)
print("Number of Elements:", temperatures_w1.size)
```

```python
#Task 3: Array Operations

# Convert Celsius to Fahrenheit

fahrenheit = (temperatures_w1 * 9/5) + 32

print("Temperatures in Fahrenheit:")
print(fahrenheit)

# Find maximum, minimum, and mean temperatures

print("Maximum Temperature:", np.max(temperatures_w1))
print("Minimum Temperature:", np.min(temperatures_w1))
print("Mean Temperature:", np.mean(temperatures_w1))
```

```python
# Task 4: Array Slicing and Indexing

# First three days
print("First Three Days:", temperatures_w1[:3])

# Weekend temperatures (last two days)
print("Weekend Temperatures:", temperatures_w1[-2:])

# Middle three days
print("Middle Three Days:", temperatures_w1[2:5])
```

```python
#Task 5: Create a 2D NumPy Array

temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])

print("Two Weeks Temperature Data:")
print(temperatures)
```

```python
#Task 6: Inspect and Slice the 2D Array

print("Shape:", temperatures.shape)
print("Data Type:", temperatures.dtype)
print("Total Number of Elements:", temperatures.size)

# Extract Week 1
print("Week 1:", temperatures[0])

# Extract Week 2
print("Week 2:", temperatures[1])

# Weekend temperatures for both weeks
print("Week 1 Weekend:", temperatures[0, -2:])
print("Week 2 Weekend:", temperatures[1, -2:])
```

** Pandas Series**

```python
#Task 1: Creating a Pandas Series

import pandas as pd

marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

print("Student Marks:")
print(marks)
```

```python
#Task 2: Indexing and Slicing


# Access first rank using integer position
print("First Rank Mark:", marks.iloc[0])

# Retrieve top 3 ranks using loc
print("Top 3 Ranks:")
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])

# Retrieve third-ranked student using iloc
print("Third Rank Mark:", marks.iloc[2])

# Boolean masking: Marks greater than 90
print("Marks Greater Than 90:")
print(marks[marks > 90])
```

```python
#Task 3: Manipulating Series

# Modify first rank mark to 100
marks.loc['Rank1'] = 100

print("After Modifying First Rank:")
print(marks)

# Remove last rank
marks = marks.drop('Rank5')

print("After Removing Last Rank:")
print(marks)

# Calculate CGPA
cgpa = marks / 10

print("CGPA:")
print(cgpa)
```

** Pandas DataFrame**

```python
#Task 1: Creating a Pandas DataFrame

data = {
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],

    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture',
        'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Furniture', 'Electronics'
    ],

    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],

    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
}

transactions = pd.DataFrame(data)

print(transactions)
```

```python
# Display the DataFrame
print(transactions)

# Display the first five rows
print(transactions.head())

# Display the last five rows
print(transactions.tail())

#Display basic information
transactions.info()

# Display the shape
print("Shape:", transactions.shape)

# Display column names
print("Column Names:", transactions.columns)

# Display data types
print("Data Types:")
print(transactions.dtypes)

# Display ProductCategory and Amount columns
print(transactions[['ProductCategory', 'Amount']])

# Retrieve the last three columns using iloc
print(transactions.iloc[:, -3:])

# Filter rows where Region is North and Amount is greater than 200
filtered_data = transactions[
    (transactions['Region'] == 'North') &
    (transactions['Amount'] > 200)]
print(filtered_data)

# Find value counts for ProductCategory
print(transactions['ProductCategory'].value_counts())

# Find unique values in Region
print("Unique Regions:")
print(transactions['Region'].unique())

# Group by Region and calculate mean Amount
region_mean = transactions.groupby('Region')['Amount'].mean()
print("Average Transaction Amount by Region:")
print(region_mean)
```

```python
#Task 3: Manipulating the DataFrame

# Modify Amount for TransactionID 102 to 165
transactions.loc[
    transactions['TransactionID'] == 102, 'Amount'
] = 165
print(transactions)

# Add a Discount column (10% of Amount)
transactions['Discount'] = transactions['Amount'] * 0.10
print(transactions)

#  Remove the row with TransactionID 109
transactions = transactions[
    transactions['TransactionID'] != 109]
print(transactions)

#  Delete the Discount column
transactions = transactions.drop(columns=['Discount'])
print("Final Transactions DataFrame:")
print(transactions)
```
